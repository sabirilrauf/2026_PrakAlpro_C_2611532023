## 2.1 Dasar Teori
Tuliskan penjelasan konsep/teori yang mendasari praktikum ini (definisi, elemen-elemen penting, dsb). Bisa memakai poin bernomor untuk merinci sub-konsep, contoh:

1. Python.<br>
Python adalah bahasa pemrograman tingkat tinggi yang dirancang dengan sintaks sederhana dan mudah dipahami. Bahasa ini mendukung berbagai paradigma pemrograman seperti prosedural, berorientasi objek, dan fungsional. Karena fleksibilitas dan kemudahan penggunaannya, Python banyak dipakai dalam pengembangan aplikasi web, analisis data, kecerdasan buatan, hingga automasi sistem. Bagi mahasiswa, Python menjadi bahasa ideal untuk mempelajari konsep algoritma dan pemrograman dasar.
2. Visual Studio Code. <br>
VS Code merupakan editor kode sumber yang dikembangkan oleh Microsoft. Editor ini ringan, cepat, dan memiliki banyak fitur seperti integrasi Git, debugging, serta dukungan ekstensi untuk berbagai bahasa pemrograman. Dengan VS Code, mahasiswa dapat menulis, menjalankan, dan mengelola kode secara lebih efisien. Selain itu, tampilan yang sederhana namun kaya fitur menjadikannya salah satu editor yang paling populer di kalangan programmer.
3. Git dan GitHub. <br>
Git adalah sistem kontrol versi yang memungkinkan pengembang melacak perubahan kode secara terstruktur. Dengan Git, setiap perubahan dapat dicatat, dikembalikan, atau digabungkan dengan mudah. GitHub adalah platform berbasis cloud yang menyediakan layanan hosting repositori Git sekaligus mendukung kolaborasi antar pengguna. Melalui GitHub, mahasiswa dapat menyimpan proyek, berbagi kode, serta berkolaborasi dalam pengembangan perangkat lunak sesuai standar industri.
