# 2.1 Dasar Teori
Perulangan for merupakan struktur dalam Python yang digunakan untuk menjalankan perintah secara berulang. Dengan perulangan for, program dapat mengulang suatu proses pada setiap elemen data atau sebanyak jumlah tertentu tanpa perlu menuliskan perintah yang sama berkali-kali.

1. Pernyataan for
Pernyataan for digunakan untuk mengulang perintah pada setiap elemen dalam suatu data, seperti list, string, atau tuple. Perulangan akan berhenti setelah seluruh elemen dalam data selesai diproses.

2. Pernyataan for dengan range()
Fungsi range() digunakan ketika jumlah perulangan sudah ditentukan. Fungsi ini menerima nilai awal, nilai akhir (tidak ikut dihitung), dan selisih antar angka. Python akan menjalankan perulangan sesuai rentang angka tersebut.

3. Nested for
Nested for digunakan ketika terdapat perulangan di dalam perulangan lain. Perulangan di dalam (inner loop) akan dijalankan sepenuhnya setiap kali perulangan di luar (outer loop) berjalan satu kali. Nested for umumnya digunakan untuk membuat pola bintang, tabel perkalian, atau mengolah data berbentuk baris dan kolom.

```python
ulang_2023 = int(input("Masukkan jumlah perulangan: "))

jumlah_2023 = 0
for i in range(1, ulang_2023 + 1):
    print(i, end=" ")
    jumlah_2023 = jumlah_2023 + i

    if i == ulang_2023:
        print(" + ", end=" ")
    else:
        print(" = ", jumlah_2023, end=" ")
print()
print("Jumlah =", jumlah_2023)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

1. Baris `ulang_2023 = int(input("Masukkan jumlah perulangan: "))` meminta pengguna memasukkan sebuah angka. Fungsi `int()` mengubah teks dari `input()` menjadi bilangan bulat, lalu disimpan pada variabel `ulang_2023`.

2. Baris `jumlah_2023 = 0` membuat variabel penampung hasil penjumlahan dengan nilai awal 0.

3. Baris `for i in range(1, ulang_2023 + 1):` membuat perulangan dari angka 1 sampai `ulang_2023`. Nilai akhir ditulis `ulang_2023 + 1` karena batas akhir `range()` tidak ikut dihitung.

4. Baris `print(i, end=" ")` menampilkan angka saat ini. Parameter `end=" "` membuat keluaran tetap berada dalam satu baris dan dipisahkan spasi.

5. Baris `jumlah_2023 = jumlah_2023 + i` menambahkan angka saat ini ke total, sehingga `jumlah_2023` terus bertambah di setiap putaran.

6. Kondisi `if i == ulang_2023:` memeriksa apakah angka saat ini adalah angka terakhir. Jika benar, program mencetak tanda `=` beserta hasil total.

7. Bagian `else:` dijalankan jika angka belum terakhir, sehingga program mencetak tanda `+` sebagai pemisah menuju angka berikutnya.

8. Baris `print("Jumlah =", jumlah_2023)` menampilkan hasil akhir penjumlahan setelah perulangan selesai.

### 2.2.4 Program nested_for1_2611532023
</div>

```python
batas_2023 = int(input("Masukkan nilai batas: "))
for line_2023 in range(1, batas_2023 + 1):
    for j_2023 in range(1, (-1 * line_2023 + batas_2023) + 1):
        print(".", end="")
    print(line_2023)
```

<div style="font-family: 'Times New Roman', Times, serif;font-size: 18px;">

Penjelasan:
1. Baris `batas_2023 = int(input("Masukkan nilai batas: "))` meminta pengguna memasukkan sebuah angka. Fungsi `int()` mengubah teks dari `input()` menjadi bilangan bulat, lalu disimpan pada variabel `batas_2023`. Nilai ini menentukan jumlah baris yang ditampilkan.

2. Baris `for line_2023 in range(1, batas_2023 + 1):` adalah perulangan luar (outer loop). Perulangan ini berjalan dari 1 sampai `batas_2023` dan mewakili nomor baris yang sedang dicetak. Nilai akhir ditulis `batas_2023 + 1` karena batas akhir `range()` tidak ikut dihitung.

3. Baris `for j_2023 in range(1, (-1 * line_2023 + batas_2023) + 1):` adalah perulangan dalam (inner loop). Ekspresi `-1 * line_2023 + batas_2023` sama dengan `batas_2023 - line_2023`, sehingga jumlah titik pada setiap baris adalah `batas_2023 - line_2023`. Semakin besar nomor baris, semakin sedikit titiknya.

4. Baris `print(".", end="")` mencetak satu titik. Parameter `end=""` membuat titik-titik tercetak berurutan dalam satu baris tanpa spasi dan tanpa pindah baris.

5. Baris `print(line_2023)` dijalankan setelah perulangan dalam selesai. Baris ini mencetak nomor baris di ujung kanan titik-titik, lalu pindah ke baris berikutnya.

```python
print("==== PROGRAM JAM PASIR KRISTAL PALINDROMIK (PEKAN 5) ====")
x_2023 = int(input("Masukkan ukuran skala jam pasir (N): "))

# ---------- Bingkai atas ----------
print("#", end="")
for garis_2023 in range(4 * x_2023 + 5):
    print("=", end="")
print("#")

# ---------- Fase 1: jam pasir atas (baris N turun s.d. 1) ----------
for baris_2023 in range(x_2023, 0, -1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_2023 in range(2 * (x_2023 - baris_2023)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_2023 in range(baris_2023, 0, -1):
        print(angka_mundur_2023, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_2023 in range(1, baris_2023 + 1):
        print(" ", end="")
        print(angka_maju_2023, end="")
    # spasi penyeimbang kanan
    for spasi_kanax_2023 in range(2 * (x_2023 - baris_2023)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Fase 2: poros titik pusat ----------
print("|", end="")
for spasi_kiri_2023 in range(2 * x_2023 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_kanax_2023 in range(2 * x_2023 + 1):
    print(" ", end="")
print("|", end="")
print()

# ---------- Fase 3: jam pasir bawah (baris 1 naik s.d. N) ----------
for baris_2023 in range(1, x_2023 + 1):
    print("|", end="")
    print(" ", end="")
    # spasi penyeimbang kiri
    for spasi_kiri_2023 in range(2 * (x_2023 - baris_2023)):
        print(" ", end="")
    # deret angka mundur (baris -> 1)
    for angka_mundur_2023 in range(baris_2023, 0, -1):
        print(angka_mundur_2023, end=" ")
    # poros kristal
    print("<*>", end="")
    # deret angka maju (1 -> baris)
    for angka_maju_2023 in range(1, baris_2023 + 1):
        print(" ", end="")
        print(angka_maju_2023, end="")
    # spasi penyeimbang kanan
    for spasi_kanax_2023 in range(2 * (x_2023 - baris_2023)):
        print(" ", end="")
    print(" ", end="")
    print("|", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for garis_2023 in range(4 * x_2023 + 5):
    print("=", end="")
print("#")
```

```python
batas_2023= int(input("Masukkan tinggi segitiga: "))
for line_2023 in range(1, batas_2023 + 1):
    for i_2023 in range(batas_2023 - line_2023) :
        print(" ", end="")
    for j_2023 in range(line_2023):
        print("*", end=" ")
    print()    
```

## PENJELASAN
1. Baris `batas_2023= int(input("Masukkan tinggi segitiga: "))` meminta pengguna memasukkan sebuah angka. Fungsi `int()` mengubah teks dari `input()` menjadi bilangan bulat, lalu disimpan pada variabel `batas_2023`. Nilai ini menentukan tinggi segitiga.

2. Baris `for line_2023 in range(1, batas_2023 + 1):` adalah perulangan luar (outer loop). Perulangan ini berjalan dari 1 sampai `batas_2023` dan mewakili nomor baris yang sedang dicetak. Nilai akhir ditulis `batas_2023 + 1` karena batas akhir `range()` tidak ikut dihitung.

3. Baris `for i_2023 in range(batas_2023 - line_2023) :` adalah perulangan dalam pertama yang mencetak spasi di sebelah kiri. Jumlah spasinya `batas_2023 - line_2023`, sehingga baris pertama memiliki spasi paling banyak dan baris terakhir tidak memiliki spasi. Inilah yang membuat bintang bergeser ke kanan.

4. Baris `print(" ", end="")` mencetak satu spasi. Parameter `end=""` membuat spasi tercetak berurutan dalam satu baris tanpa pindah baris.

5. Baris `for j_2023 in range(line_2023):` adalah perulangan dalam kedua yang mencetak bintang. Jumlah bintangnya sama dengan nomor baris: baris 1 memiliki 1 bintang, baris 2 memiliki 2 bintang, dan seterusnya.

6. Baris `print("*", end=" ")` mencetak satu bintang diikuti spasi. Parameter `end=" "` membuat bintang-bintang dipisahkan spasi dan tetap berada dalam satu baris.

7. Baris `print()` dijalankan setelah kedua perulangan dalam selesai. Baris ini hanya berfungsi pindah ke baris baru, sehingga baris berikutnya dicetak di bawahnya.
